```python

if not MANIFEST_PATH.exists():
    raise FileNotFoundError(f"Không tìm thấy file manifest.csv tại: {MANIFEST_PATH}")

manifest_df = pd.read_csv(MANIFEST_PATH)
print("--> Các cột hiện có trong manifest.csv:", list(manifest_df.columns))
print(f"--> Tổng số dòng trong manifest.csv: {len(manifest_df)}")

possible_cols = ['file_path', 'source_file', 'file_name', 'filename', 'filepath', 'path', 'file', 'pdf_file', 'pdf_name']
file_col = None
for col in manifest_df.columns:
    if col.lower().strip() in possible_cols:
        file_col = col
        break

if not file_col:
    for col in manifest_df.columns:
        sample_vals = manifest_df[col].dropna().astype(str).tolist()
        if any(v.lower().endswith('.pdf') for v in sample_vals):
            file_col = col
            break

if not file_col:
    raise KeyError(f"Không thể tự xác định cột chứa tên file PDF! Các cột hiện có: {list(manifest_df.columns)}")

print(f"--> Đã chọn cột chứa file: '{file_col}'")

with open(LEXICON_PATH, 'r', encoding='utf-8') as f:
    lexicon = json.load(f)
stop_phrases = lexicon.get("stop_phrases_to_remove_first", [])

results = []

for idx, row in manifest_df.iterrows():
    val = row[file_col]
    if pd.isna(val):
        continue  
        
    file_rel_path = str(val).strip()
    pdf_path = ROOT / file_rel_path
    
    if not pdf_path.exists():
        matched = list(ROOT.rglob(Path(file_rel_path).name))
        if matched:
            pdf_path = matched[0]
        else:
            print(f"[BỎ QUA] Không tìm thấy file: {file_rel_path}")
            continue

    print(f"[{idx+1}/{len(manifest_df)}] Đang xử lý: {pdf_path.name}...")
    
    
    file_hash = calculate_md5(pdf_path)

    with pymupdf.open(pdf_path) as doc:
        raw_text = '\n\n'.join([p.get_text('text') or '' for p in doc])
        total_pages = len(doc)

    clean_text = clean_bank_text(raw_text, stop_phrases)
    total_words = len(clean_text.split())
    
    base_name = pdf_path.stem
    with open(EXTRACTS_DIR / f"{base_name}.txt", "w", encoding="utf-8") as f:
        f.write(raw_text)
    with open(MIXED_DIR / f"{base_name}_clean.txt", "w", encoding="utf-8") as f:
        f.write(clean_text)

    pillar_counts = count_pillar_frequencies(clean_text, lexicon)

    bank_val = row.get('bank') or row.get('bank_name') or row.get('ticker') or 'UNKNOWN'
    year_val = row.get('fiscal_year') or row.get('year') or row.get('nam') or 'UNKNOWN'
    type_val = row.get('report_type') or row.get('type') or row.get('loai_bao_cao') or 'UNKNOWN'

    record = {
        'file_name': pdf_path.name,
        'file_hash_md5': file_hash,
        'bank': str(bank_val).upper(),
        'fiscal_year': year_val,
        'report_type': type_val,
        'total_pages': total_pages,
        'total_words': total_words,
        'total_dx_hits': sum(pillar_counts.values())
    }
    record.update(pillar_counts)
    results.append(record)
```

```python
import pandas as pd
final_df = pd.DataFrame(results)

final_df['total_words'] = final_df['total_words'].replace(0, 1)
for col in [c for c in final_df.columns if c.startswith('count_')]:
    pillar = col.replace('count_', '')
    final_df[f'norm_freq_{pillar}'] = (final_df[col] / final_df['total_words']) * 10000

FINAL_CSV = OUT_DIR / "dbi_final_truth.csv"
FINAL_XLSX = OUT_DIR / "dbi_final_truth.xlsx"

final_df.to_csv(FINAL_CSV, index=False, encoding='utf-8-sig')
final_df.to_excel(FINAL_XLSX, index=False)

print("="*60)
print("🎉 ĐÃ XUẤT FILE THÀNH CÔNG!")
print(f"1. File CSV: {FINAL_CSV}")
print(f"2. File Excel: {FINAL_XLSX}")
print("="*60)
display(final_df.head())
```

```python
import pandas as pd

df_master = pd.read_excel(OUT_DIR / "dbi_audit_master_237_files.xlsx")

def detect_report_type(fname):
    name_lower = str(fname).lower()
    if 'sustainability' in name_lower or 'esg' in name_lower or 'ptbv' in name_lower:
        return 'sustainability'
    elif 'annual' in name_lower or 'bctn' in name_lower or 'thuong_nien' in name_lower:
        return 'annual_reports'
    elif 'financial' in name_lower or 'bctc' in name_lower:
        return 'financial_statements'
    return 'other'

df_master['report_type'] = df_master['file_name'].apply(detect_report_type)

df_master.to_excel(OUT_DIR / "dbi_audit_master_237_files.xlsx", index=False)

clean_df = df_master[
    (df_master['total_words'] > 3000) & 
    (df_master['report_type'].isin(['annual_reports', 'sustainability']))
].copy()

clean_df = clean_df.sort_values(by=['bank', 'fiscal_year', 'report_type'])

SCORING_FILE = OUT_DIR / "dbi_clean_for_scoring.xlsx"
clean_df.to_excel(SCORING_FILE, index=False)

cols_count = [c for c in clean_df.columns if c.startswith('count_')]
summary_df = clean_df.groupby(['bank', 'fiscal_year'], as_index=False)[cols_count + ['total_words', 'total_dx_hits']].sum()

for col in cols_count:
    pillar = col.replace('count_', '')
    summary_df[f'norm_freq_{pillar}'] = (summary_df[col] / summary_df['total_words']) * 10000

SUMMARY_FILE = OUT_DIR / "dbi_final_for_step4_scoring.xlsx"
summary_df.to_excel(SUMMARY_FILE, index=False)

print("="*60)
print(f"🎉 ĐÃ SỬA XONG TOÀN BỘ!")
print(f"1. File Clean từng báo cáo (không còn trắng): {SCORING_FILE.name} ({len(clean_df)} dòng)")
print(f"2. File Tổng hợp gộp theo Ngân hàng - Năm (Nộp Bước 4): {SUMMARY_FILE.name} ({len(summary_df)} dòng)")
print("="*60)
display(summary_df)
```
